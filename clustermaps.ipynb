```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
import scipy.stats as stats
```

```python
df=pd.read_csv("noag_cpepo_DGE_kipes_selected_uvn_hn.csv",delimiter=",",index_col=0,header=0)
annotations = df.iloc[:,1]
expr = df.iloc[:, 7:17].astype("float_")
df_log = np.log1p(expr)
top_genes = df_log.var(axis=1).nlargest(200).index
top_df = df_log.loc[top_genes]
top_df.index = annotations[top_df.index]
g = sns.clustermap(top_df, xticklabels=True, figsize=(12, 12), z_score=0, col_cluster=False)
plt.savefig("uvn_hn_cpepo_phenylpropanoid_heatmap.png")
```

```python
df=pd.read_csv("noag_fvesca_DGE_kipes_selected_uvn_hn.csv",delimiter="\t",index_col=0,header=0)
annotations = df.iloc[:,1]
expr = df.iloc[:, 7:17].astype("float_")
df_log = np.log1p(expr)
top_genes = df_log.var(axis=1).nlargest(200).index
top_df = df_log.loc[top_genes]
top_df.index = annotations[top_df.index]
g = sns.clustermap(top_df, xticklabels=True, figsize=(12, 6), z_score=0, col_cluster=False)
plt.setp(g.ax_heatmap.get_yticklabels(), rotation=0)
plt.savefig("uvn_hn_fvesca_phenylpropanoid_heatmap.png")
```
